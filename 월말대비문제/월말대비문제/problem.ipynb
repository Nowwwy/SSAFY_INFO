```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# Python 내장 함수 sum 을 사용하지 않으면 추가 점수
def sum_list(numbers):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - numbers: 정수로 이루어진 리스트
    total = 0
    for number in numbers:
        total += number
    return total


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(sum_list([10, 20, 30]))  # 60
print(sum_list([0, 0, 0]))  # 0
print(sum_list([5, 15, 25, 55]))  # 100
print(sum_list([100]))  # 100
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# 문자열 메서드 count 를 사용하지 않으면 추가 점수
def count_char(word, target):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - word: 문자열 / target: 세고 싶은 한 글자(문자열)
    count = 0
    for i in word:
        if i == target:
            count += 1
    return count


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(count_char('banana', 'a'))  # 3
print(count_char('banana', 'b'))  # 1
print(count_char('ssafy', 's'))  # 2
print(count_char('python', 'z'))  # 0
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def highest_scorer(scores):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - scores: {이름(str): 점수(int)} 형태의 딕셔너리 (빈 딕셔너리는 들어오지 않음)
    highest_name = ''
    highest_score = 0

    for name in scores:
        if scores[name] > highest_score:
            highest_score = scores[name]
            highest_name = name
    return highest_name



# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(highest_scorer({'kim': 90, 'lee': 85, 'park': 95}))  # park
print(highest_scorer({'a': 70, 'b': 70, 'c': 60}))  # a
print(highest_scorer({'solo': 100}))  # solo
print(highest_scorer({'x': 30, 'y': 80, 'z': 80}))  # y
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# 슬라이싱을 이용해 뒤집은 문자열과 비교하면 간단하다
def is_palindrome(word):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - word: 문자열
    if word == word[::-1]:
        return True
    return False


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(is_palindrome('level'))  # True
print(is_palindrome('python'))  # False
print(is_palindrome('noon'))  # True
print(is_palindrome('a'))  # True
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# 세트의 교집합(& 또는 intersection)을 활용하면 편하다
def common_items(list1, list2):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - list1, list2: 값이 담긴 리스트
    return sorted(set(list1) & set(list2))


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(common_items([1, 2, 3, 4], [2, 4, 6]))  # [2, 4]
print(common_items(['a', 'b', 'c'], ['b', 'c', 'd']))  # ['b', 'c']
print(common_items([1, 2, 3], [4, 5, 6]))  # []
print(common_items([5, 5, 6, 7], [7, 6, 6]))  # [6, 7]
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def longest_word(sentence):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - sentence: 공백으로 단어가 구분된 문자열 (빈 문자열은 들어오지 않음)
    words = sentence.split()
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest



# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(longest_word('i love python programming'))  # programming
print(longest_word('a bb ccc dd'))  # ccc
print(longest_word('hello world'))  # hello
print(longest_word('ssafy'))  # ssafy
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# 리스트의 count 메서드를 쓰지 않고 딕셔너리로 직접 세면 더 좋다
def count_frequency(items):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - items: 값이 담긴 리스트
    result = {}

    for item in items:
        if item in result:
            result[item] += 1
        else:
            result[item] = 1
    return result

# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(count_frequency(['a', 'b', 'a', 'c', 'b', 'a']))  # {'a': 3, 'b': 2, 'c': 1}
print(count_frequency([1, 1, 1, 2]))  # {1: 3, 2: 1}
print(count_frequency(['x']))  # {'x': 1}
print(count_frequency([]))  # {}
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def group_by_length(words):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - words: 단어(str) 리스트
    result = {}

    for word in words:
        length = len(word)

        if length not in result:
            result[length] = []

        result[length].append(word)

    return result


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(group_by_length(['sky', 'sea', 'tree', 'a', 'go']))  # {3: ['sky', 'sea'], 4: ['tree'], 1: ['a'], 2: ['go']}
print(group_by_length(['cat', 'dog', 'bird']))  # {3: ['cat', 'dog'], 4: ['bird']}
print(group_by_length(['one']))  # {3: ['one']}
print(group_by_length([]))  # {}
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def first_unique_char(word):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - word: 문자열
    alone = {}

    for char in word:
        if char in alone:
            alone[char] += 1
        else:
            alone[char] = 1

    for char in word:
        if alone[char] == 1:
            return char
    return ''


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(first_unique_char('swiss'))  # w
print(first_unique_char('aabbcc'))  # 
print(first_unique_char('python'))  # p
print(first_unique_char('aabbc'))  # c
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def merge_stock(stock1, stock2):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - stock1, stock2: {상품명(str): 수량(int)} 딕셔너리
    result = stock1.copy()

    for item in stock2:
        if item in result:
            result[item] += stock2[item]
        else:
            result[item] = stock2[item]
    return result

# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(merge_stock({'apple': 3, 'banana': 2}, {'banana': 5, 'cherry': 1}))  # {'apple': 3, 'banana': 7, 'cherry': 1}
print(merge_stock({'a': 1}, {'b': 2}))  # {'a': 1, 'b': 2}
print(merge_stock({'x': 10}, {'x': 5}))  # {'x': 15}
print(merge_stock({}, {'new': 7}))  # {'new': 7}
#####################################################

# result[item] = result.get(item, 0) + stock2[item]
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
# user_id[:3] 슬라이싱과 '*' * 개수 를 조합해 보라
def mask_id(user_id):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - user_id: 문자열
    first = user_id[:3]
    count = len(user_id) - 3
    stars = '*' * count
    return first + stars


# return user_id[:3] + '*' * max(0, len(user_id) - 3)


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(mask_id('jungssafy'))  # jun******
print(mask_id('kimssafy'))  # kim*****
print(mask_id('abc'))  # abc
print(mask_id('ab'))  # ab
#####################################################
```

```python
############## 주의 ##############
# 입력을 받기 위한 input 함수는 절대 사용하지 않습니다.
def run_length_encode(word):
    # 여기에 코드를 작성하여 함수를 완성합니다.
    # 입력 - word: 문자열
    if not word:
        return ''

    result = ''
    count = 1

    for i in range(1, len(word)):
        if word[i] == word[i - 1]:
            count += 1
        else:
            result += word[i - 1] + str(count)
            count = 1

    result += word[-1] + str(count)

    return result


# 추가 테스트를 위한 코드 작성 가능
# 예) print(함수명(인자))

#####################################################
# 아래 코드를 삭제하는 경우
# 모든 책임은 삭제한 본인에게 있습니다.
############## 테스트 코드 삭제 금지 #################
print(run_length_encode('aaabbc'))  # a3b2c1
print(run_length_encode('abc'))  # a1b1c1
print(run_length_encode('aaaa'))  # a4
print(run_length_encode(''))  # 
#####################################################
```
