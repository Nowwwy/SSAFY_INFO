### RNN으로 데이터 학습 원리

- RNN은 이전 시점의 Hidden State의 출력값이, 다음 Hidden State 계산의 입력값으로 들어갑니다.
- RNN의 학습은 다음과 같이 이뤄집니다.
  - **이전 데이터들로 만들어진 정보(Hidden Layer 출력값)를, 다음 글자를 예측하는데 사용됩니다.**

  - 학습할때 입력데이터로 **['안', '녕, '하', '세, '요]** 를 넣어서 학습하겠다고 하겠습니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/05.png)

- 나름대로 '안' 다음 글자를 'K'로, 예측했고.. 이런식으로 각각 다음글자를 ['K', '호', '커'] 로 예측이 잘 안되었습니다.
- 이후에 GD를 통해, 오차율이 적도록 히든레이어 가중치 값이 업데이트 됩니다.
  - 그러면 점점 예측을 잘하게 될 것입니다.

- 아래 그림은 epoch을 반복해서 점점 똑똑해졌을 때.. 학습하고 있는 이미지입니다.
  - '녕' 까지는 예측이 잘 되었고, '!' 랑 'ㅋ' 는 오류가 있네요.

    - 예측해보고, GD로 점점 에러가적도록 파라미터를 업데이트 합니다.
    - 이후에 실제로 사용할때는 RNN이 추론한 글자들을 출력하면 됩니다.

### RNN의 단점


- Sequence가 길어질수록 초기 정보(Hidden State)들이 반영이 잘안됩니다.
  - 과거 기억을 잊게된다로 비유할 수 있겠습니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/09.png)

- 또한 Gradient Descent를 하는데 앞 부분은 학습이 잘 되는데, 뒷 부분은 학습이 잘 안됩니다.
  - GD를 위해 기울기 계산 (Backpropagation)을 합니다.
  - 그때 기울기 소실 (점점 0으로 수렴) 문제나, 기울기 폭발 (점점 거대한 값으로 폭발) 문제가 있습니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/10.png)

- 기울기 소실, 폭발문제 해결하기 위해 다른알고리즘들(LSTM)이 나왔지만, 여전히 오래전 정보를 잊어버리는 문제는 존재합니다.

## Transformer 모델에서 RNN 단점을 근본적으로 해결!


- 이러한 문제는 **Transformer 모델** 에서 근본적으로 해결되었습니다
- Transformer는 RNN처럼 과거 정보를, 다음 Layer로 전달해주는 방식을 버렸어요.
  - 참고로 RNN은 이전 Hidden Layer값이 모든 과거 정보를 내포하고 있었죠.
  - Transformer 모델은 입력값들의 정보들을 모두 참고하여 다음 Layer의 출력값을 만들어냅니다!
    - 장기기억 소실문제가 해결되었습니다.
    - 기울기 소실, 폭발 문제가 해결되었습니다.
    - 현대 딥러닝은 대부분 Transformer 모델을 기반으로 사용합니다




# 2. Attention과 Transformer 모델
- 시계열 모델(RNN 등) 한계를 극복하기 위해 탄생했습니다.
- Transformer는 Attention을 여러개 사용하는 모델입니다.
  - 일단 Attention이 무엇인지부터 보시죠.



### Attention
- 각 단어가 다른 단어들과 유사한 정도를 계산합니다.
- 이 유사성 정보들을 합쳐서, "문맥 표현 공간"의 벡터값을 계산합니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/15_easy/3_2_llm_evolution/10.png)

$\huge X_{임베딩들} \xrightarrow{\text{Attention}} Y_{\underline{\color{red}{문맥표현공간}}}$



- 더 정확히 말하면 다음과 같습니다.
  - 그 동안 일반적인 선형계산에는 우리는 y = WX + b로 계산했죠?
    - 하나의 Weight 행렬을 사용합니다.
  - Attention을 계산할 때는 3개의 Weight 행렬을 사용합니다. 3개는 이름이 있습니다.
    - Query / Key / Value 행렬 입니다.
    - 이 Weight값은 학습 대상이 됩니다~.

<br/>

- 그럼 아래 내용을 봅시다.
  - 복잡한 수식은 건강상 모자이크 처리 했습니다. ^^
  - 각 단어별 "의미공간"의 벡터값인 임베딩 벡터값을 구합니다.
  - 그리고 임베딩 벡터 값들을 각각 $X \cdot W_{query}$, $X \cdot W_{key}$, $X \cdot W_{value}$ 값을 구합니다.
  - 이후에 Attention 수식에 대입하면 output을 구할 수 있습니다.
  - 이 output은 "문맥표현공간"의 벡터값이 됩니다.
    - 엄청난 고차원의 문맥 세계의 입주민이 되는겁니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/11.png)

위의 식을 통해서 **나는 학교에 간다** 라는 문장을 통해 문맥표현공간으로 표현하게 된다면

## 각 토큰의 문맥 벡터 계산

- **"나는"의 문맥 벡터** : "나는"의 의미 × 0.6 + "학교에"의 의미 × 0.1 + "간다"의 의미 × 0.3
- **"학교에"의 문맥 벡터** : "나는"의 의미 × 0.2 + "학교에"의 의미 × 0.5 + "간다"의 의미 × 0.3
- **"간다"의 문맥 벡터** : "나는"의 의미 × 0.3 + "학교에"의 의미 × 0.2 + "간다"의 의미 × 0.5

예를 들어 "학교에"라는 단어의 문맥 벡터는:
- "나는"의 의미 × 0.2 + "학교에"의 의미 × 0.5 + "간다"의 의미 × 0.3

이렇게 계산되어 단순한 <u>"학교"라는 단어 뜻이 아닌 **"누가 어디로 가는지"라는 문맥이 반영된 의미**가 됩니다</u>.
<br/><br/>
이렇게 어텐션을 통해 계산된 개별 문맥 벡터 하나하나가 모여 존재하는 벡터 공간을 **문맥 표현 공간**이라고 합니다.

<br/>
(아래 3D 그래프는 극적적인 시각화를 위해 3차원 임베딩 벡터를 사용했습니다.)

```python
# @title
import plotly.graph_objects as go
import numpy as np

# 토큰 이름
tokens = ['나는', '학교에', '간다']

# 각 토큰의 의미 벡터 (3D)
semantic_vectors = np.array([
    [0.8, 0.3, 0.4],   # "나는"
    [0.3, 0.9, 0.5],   # "학교에"
    [0.5, 0.5, 0.8]    # "간다"
])

# 어텐션 가중치 행렬
attention_weights = np.array([
    [0.6, 0.1, 0.3],  # "나는"이 각 토큰에 주는 가중치
    [0.2, 0.5, 0.3],  # "학교에"가 각 토큰에 주는 가중치
    [0.3, 0.2, 0.5]   # "간다"가 각 토큰에 주는 가중치
])

# 문맥 벡터 계산
context_vectors = attention_weights @ semantic_vectors

# print("=" * 60)
# print("각 토큰의 문맥 벡터 계산 결과")
# print("=" * 60)
# for i, token in enumerate(tokens):
#     print(f'\n"{token}"의 문맥 벡터:')
#     print(f'  = {attention_weights[i][0]:.1f} × "나는"의 의미 + {attention_weights[i][1]:.1f} × "학교에"의 의미 + {attention_weights[i][2]:.1f} × "간다"의 의미')
#     print(f'  = {attention_weights[i][0]:.1f} × {semantic_vectors[0]} + {attention_weights[i][1]:.1f} × {semantic_vectors[1]} + {attention_weights[i][2]:.1f} × {semantic_vectors[2]}')
#     print(f'  = {context_vectors[i]}')
# print("\n" + "=" * 60)

# Plotly Figure 생성
fig = go.Figure()

# 색상 정의
colors = ['#4f46e5', '#06b6d4', '#10b981']  # 인디고, 시안, 에메랄드
semantic_color = '#94a3b8'  # 회색

# 1. 원점 표시
fig.add_trace(go.Scatter3d(
    x=[0], y=[0], z=[0],
    mode='markers',
    marker=dict(size=8, color='black'),
    name='원점',
    showlegend=True
))

# 2. 의미 벡터 표시 (반투명)
for i, token in enumerate(tokens):
    vec = semantic_vectors[i]

    # 벡터 화살표 (선)
    fig.add_trace(go.Scatter3d(
        x=[0, vec[0]],
        y=[0, vec[1]],
        z=[0, vec[2]],
        mode='lines+markers+text',
        line=dict(color=semantic_color, width=4, dash='dash'),
        marker=dict(size=[0, 6], color=semantic_color),
        text=['', f'{token}\n(의미)'],
        textposition='top center',
        textfont=dict(size=10, color=semantic_color),
        name=f'{token} (의미 벡터)',
        showlegend=True,
        opacity=0.4
    ))

# 3. 문맥 벡터 표시 (강조)
for i, token in enumerate(tokens):
    vec = context_vectors[i]

    # 벡터 화살표
    fig.add_trace(go.Scatter3d(
        x=[0, vec[0]],
        y=[0, vec[1]],
        z=[0, vec[2]],
        mode='lines+markers+text',
        line=dict(color=colors[i], width=8),
        marker=dict(size=[0, 10], color=colors[i], symbol='diamond'),
        text=['', f'{token}\n(문맥)'],
        textposition='top center',
        textfont=dict(size=12, color=colors[i], family='Arial Black'),
        name=f'{token} (문맥 벡터)',
        showlegend=True
    ))

# 4. 화살표 cone 추가 (문맥 벡터용)
for i, token in enumerate(tokens):
    vec = context_vectors[i]
    # 벡터의 끝점에서 방향을 나타내는 작은 cone
    fig.add_trace(go.Cone(
        x=[vec[0]],
        y=[vec[1]],
        z=[vec[2]],
        u=[vec[0]*0.1],
        v=[vec[1]*0.1],
        w=[vec[2]*0.1],
        colorscale=[[0, colors[i]], [1, colors[i]]],
        showscale=False,
        sizemode='absolute',
        sizeref=0.1,
        name=f'{token} 방향',
        showlegend=False
    ))

# 레이아웃 설정
fig.update_layout(
    title=dict(
        text='어텐션 메커니즘: 3D 문맥 벡터 시각화<br><sub>의미 벡터(점선) → 문맥 벡터(실선)</sub>',
        x=0.5,
        xanchor='center',
        font=dict(size=20)
    ),
    scene=dict(
        xaxis=dict(title='차원 1 (X)', range=[-0.1, 1.0], gridcolor='lightgray'),
        yaxis=dict(title='차원 2 (Y)', range=[-0.1, 1.0], gridcolor='lightgray'),
        zaxis=dict(title='차원 3 (Z)', range=[-0.1, 1.0], gridcolor='lightgray'),
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.2)
        ),
        bgcolor='rgb(250, 250, 250)'
    ),
    width=1000,
    height=800,
    showlegend=True,
    legend=dict(
        x=0.02,
        y=0.98,
        bgcolor='rgba(255, 255, 255, 0.8)',
        bordercolor='gray',
        borderwidth=1
    )
)
fig.show()
```

- 위 내용을 요약해보겠습니다.
- 먼저 아래와 같이 Q, K, V 는 각각의 Weight값을 곱한값입니다.
  - 임베딩 값에 선형변환을 한겁니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/15_easy/3_2_llm_evolution/12.png)

$\huge
\begin{align}
Q &= X \cdot W_{query} \\
K &= X \cdot W_{key} \\
V &= X \cdot W_{value}
\end{align}
$

- 이후에 수식은 다음과 같이 요약할 수 있습니다.
  - Attention 함수는 아까 모자이크 했던 수식입니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/15_easy/3_2_llm_evolution/13.png)

$\huge Y_{문맥표현공간} = \color{red}{f_{attention}(Q, K, V)}$

### Transformer 구조
- Transformer의 구조는 아래와 같습니다.
  - 아래 이미지 출처 : https://en.wikipedia.org/wiki/Transformer_(deep_learning_architecture)

- Tranformer는 Encoder들 + Decoder들 조합으로 구성되어 있습니다.
  - Encoder, Decorder 모두 Attention과 MLP의 조합으로 구성되어 있습니다.
  - Encoder : 문맥을 이해합니다.
  - Decoder : 입력 문장 정보를 이용해서, 출력 문장을 생성합니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/12.png)

- 요약하면 다음과 같습니다.
  - 입력 문장 --> [Encoder × N개] --> 문맥 벡터
    - Encoder 여러개로 정교한 이해를 합니다.
  - 출력 문장 --> [Decoder × N개] --> 최종 문장 예측
    - Decoder 여러개로 점차 정교한 문장을 만들어냅니다.
    - 입력 : Decoder는 Encoder의 문맥 정보(State)들과 본인이 출력한 단어가 입력됩니다.
    - 출력 : Softmax 값으로, Vocabulary 들 각각 단어가 나올 확률들이 나옵니다.

# 3. Transformer 모델을 응용하여 만들 수 있는 것들

### 모달이란?
  - 인간이 정보를 인식할 수 있는 형태를 모달(Modality) 이라고 합니다.
    - Text, Image, Audio, .. 여러 Sensor 값들이 될 수 있습니다.
    - 여러 모달을 다루는 모델을, **멀티모달 모델**이라고 합니다.

### 모달별 Transformer 활용 원리

- Text 모달

# 5. Transformer 기반, 사물인식 모델


- 이미지에서 미리 학습된 객체들을 탐지하는 모델을 돌려보겠습니다.
  - Object Detection 모델이라고 합니다.
- 이제 GPU를 켜주세요.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/20.png)

- 사용하는 모델은 Transformer로 만들어진 YOLOS 모델입니다.
  - CNN 기반 모델로 유명한 YOLO 모델은 (You Only Look Once) 이고요.
  - 지금 사용할 Transformer 기반 YOLOS 모델은 (You Only Look at One Sequence) 입니다.
    - 이름을 패러디한거예요.
    - hustvl/yolos-small (https://huggingface.co/hustvl/yolos-small)

  - 먼저 아래 명령어를 수행해서 공개 폰트를 다운로드받아주세요.
    - 이미지에 사물인식 결과를 표시하는데, 폰트가 너무 작아요.
    - 폰트사이즈를 키우려면 따로 폰트파일이 필요하다고 합니다. ^^;

```python
import urllib.request

url = "https://github.com/google/fonts/raw/main/ofl/jejugothic/JejuGothic-Regular.ttf"
urllib.request.urlretrieve(url, "JejuGothic-Regular.ttf")

# 참고: Linux 환경에서는 아래 명령어로도 받을 수 있어요.
#!wget https://github.com/google/fonts/raw/main/ofl/jejugothic/JejuGothic-Regular.ttf
```

```python
# CPU에서 57초 걸림
from transformers import YolosImageProcessor, YolosForObjectDetection
import requests
import torch
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt

# 이미지 로드 후 전처리
url = 'https://cdn.pixabay.com/photo/2021/07/09/02/29/family-6398107_1280.jpg'
image = Image.open(requests.get(url, stream=True).raw)
image = image.resize((640, 462))

# 전처리기, 모델 로드
feature_extractor = YolosImageProcessor.from_pretrained('hustvl/yolos-small')
model = YolosForObjectDetection.from_pretrained('hustvl/yolos-small')

# 입력 전처리
inputs = feature_extractor(images=image, return_tensors="pt")

# 모델추론
outputs = model(**inputs)

# 후처리
target_sizes = torch.tensor([image.size[::-1]])
results = feature_extractor.post_process_object_detection(
    outputs, threshold=0.7, target_sizes=target_sizes
)[0]

# 이미지에 바운딩박스 그리기
draw = ImageDraw.Draw(image)
font = ImageFont.truetype("JejuGothic-Regular.ttf", 16)

for score, label, box in zip(results["scores"], results["labels"], results["boxes"]):
    box = [round(i, 2) for i in box.tolist()]
    label_text = f"{model.config.id2label[label.item()]} {round(score.item(), 2)}"

    draw.rectangle(box, outline="red", width=2) # 빨간 사각형
    draw.text((box[0], box[1] - 20), label_text, fill="red", font=font) # 빨간색 글씨

# 결과 시각화
plt.figure(figsize=(10, 10))
plt.imshow(image)
plt.axis("off")
plt.show()
```

# 6. Transformer 기반, 이미지 생성 멀티모달 모델
- Text to Image 모델로, 이미지를 생성해봅시다.
  - 입력은 Text이고, 출력은 Image 입니다. 두 가지 모달을 사용했으니 멀티모달 모델입니다~.
- 이미지 생성방법인 **Diffusion 알고리즘**은 다음과 같이 동작됩니다.
  - 학습 : 원본 이미지에 노이즈를 조금 추가하고 어디에 얼마나 노이즈를 추가했는지 예측합니다. 이를 반복합니다.
  - 추론 : 완전한 노이즈로 시작해서, 노이즈를 점점 제거하면서 복원하는 방법으로 이미지를 생성합니다.

- 이미지 출처 : https://developer.nvidia.com/blog/understanding-diffusion-models-an-essential-guide-for-aec-professionals/

![df.gif](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_2_llm_evolution/21.gif)

- 다음과 같은 모델을 사용했습니다.
  - "stabilityai/sd-turbo 모델입니다. (Stable Diffusion 2.1)
    - https://huggingface.co/stabilityai/sd-turbo
    - 적은 노이즈 제거 Step으로 빠르게 이미지 생성이 가능한 모델입니다.

```python
# 로컬에서는 설치 필요. 최초 1번 실행 후 커널 재시작
#!pip install diffusers==0.37.0 accelerate==1.13.0

# CPU로 실행 안됨: 아래와 같은 오류가 발생하니 GPU로 변경 후 실행
# 약 2분 걸림
# RuntimeError: Found no NVIDIA driver on your system.
# Please check that you have an NVIDIA GPU and installed a driver from http://www.nvidia.com/Download/index.aspx
import torch
from diffusers import AutoPipelineForText2Image
import matplotlib.pyplot as plt

# 1. 모델 로드 (RTX 5060 및 최신 라이브러리 최적화 설정)
pipe = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sd-turbo",
    torch_dtype=torch.bfloat16,      # Blackwell 아키텍처는 BF16 성능이 매우 강력합니다.
    variant="fp16",                  # 저장 공간 및 로딩 속도 절약
    use_safetensors=True,
    low_cpu_mem_usage=True           # accelerate 라이브러리 활용
)

# 2. GPU(CUDA)로 이동
pipe.to("cuda")

# 프롬프트 입력
prompt = "The puppy is eating a hamburger at McDonald's."

# 3. [핵심] PyTorch 2.8.0 컴파일 최적화
# 첫 실행 시 1~2분 정도 소요되지만, 이후 생성 속도가 약 20% 이상 빨라집니다.
pipe.unet = torch.compile(pipe.unet, mode="reduce-overhead")

# 4. 이미지 생성 테스트
image = pipe(prompt=prompt, num_inference_steps=2, guidance_scale=0.0).images[0]

# 5. 결과 확인 1
image.show()

# 5. 결과 확인 2
# plt.figure(figsize=(8, 8))
# plt.imshow(image)
# plt.axis("off")
# plt.show()

# 저장
image.save("result.png")
```

### 생성 이미지 평가하기: CLIP score

- 방금 생성한 이미지, 프롬프트대로 잘 만들어졌을까요? 눈으로만 보지 말고 **숫자로** 평가해봅시다.
- 이때 사용하는 모델이 **CLIP** 입니다.
  - 앞의 "모달별 Transformer 활용 원리"에서 배운, Text와 Image를 같은 "문맥표현공간"의 벡터로 만드는 원리로 학습된 모델입니다.
  - 이미지 벡터와 문장 벡터의 유사도 점수를 **CLIP score** 라고 부릅니다.
    - 점수가 높을수록, 이미지가 그 문장과 잘 어울린다는 뜻입니다~.
- 평가 방법은 간단합니다.
  - 정답 문장(이미지 생성에 사용한 프롬프트)과, 헷갈릴 만한 오답 문장 몇 개를 함께 넣습니다.
  - CLIP이 이미지와 각 문장의 유사도를 계산해서, 가장 잘 어울리는 문장을 골라줍니다.
  - 정답 문장이 1등이면? 프롬프트에 충실한 이미지가 생성된 것입니다.

```python
# CLIP으로 생성 이미지 평가하기 (다운로드 약 600MB, CPU에서도 몇 초면 됩니다)
import torch
from PIL import Image
from transformers import CLIPProcessor, CLIPModel

device = "cuda" if torch.cuda.is_available() else "cpu"

# 위에서 생성해 저장한 이미지를 불러옵니다.
image = Image.open("result.png")

# 평가에 사용할 문장들. 첫 문장이 이미지 생성에 사용한 프롬프트(정답)입니다.
labels = [
    "The puppy is eating a hamburger at McDonald's.",  # 정답 (생성 프롬프트)
    "The puppy is sleeping on the sofa.",              # 오답 후보 1 (행동이 다름)
    "The cat is eating a hamburger at McDonald's.",    # 오답 후보 2 (동물이 다름)
    "A photo of an empty restaurant.",                 # 오답 후보 3 (강아지가 없음)
]

# CLIP 전처리기, 모델 로드 (정규 실습과 같은 모델입니다)
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)

# 이미지와 문장들을 한 번에 전처리
inputs = processor(text=labels, images=image, return_tensors="pt", padding=True).to(device)

# 이미지-문장 유사도(CLIP score) 계산
with torch.no_grad():
    outputs = clip_model(**inputs)
    logits_per_image = outputs.logits_per_image  # 문장별 유사도 점수 (CLIP score)
    probs = logits_per_image.softmax(dim=1)      # 확률로 변환 (합계 = 1)

# 문장별 점수 출력
for label, score, prob in zip(labels, logits_per_image[0], probs[0]):
    print(f"CLIP score {score.item():6.2f} ({prob.item()*100:5.1f}%) | {label}")

# 가장 잘 어울리는 문장 선택
best_idx = probs.argmax(dim=1).item()
print(f'\nCLIP의 선택: "{labels[best_idx]}"')
```

- 결과를 해석해봅시다.
  - 정답 문장(생성 프롬프트)의 확률이 가장 높다면, 생성 이미지가 **프롬프트에 충실**하다는 뜻입니다.
  - 오답 문장이 1등이라면? 프롬프트를 더 구체적으로 바꿔서 다시 생성하고, 점수가 오르는지 비교해보세요.
- CLIP은 평가 문장을 우리가 자유롭게 정할 수 있다는 것이 큰 장점입니다.
  - 반면 3-1에서 배운 CNN 분류 모델(예: ResNet-50)은 미리 정해진 클래스(ImageNet 1000개)로만 분류할 수 있습니다.
  - 정규 실습에서는 같은 이미지를 CLIP과 ResNet-50에 넣어서, 두 모델의 차이를 직접 비교해봅니다. 기대해주세요~!

- 아래 예제는 2개의 햄버거 사진과 프롬프트를 주고, 대답을 기다립니다.
  - 프롬프트 : "Can you explain the difference between the two images?

```python
import torch
from PIL import Image
from transformers import AutoProcessor
from transformers import AutoModelForImageTextToText
from transformers.image_utils import load_image
import matplotlib.pyplot as plt

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Load images & print
image1 = load_image("https://cdn.pixabay.com/photo/2022/07/15/18/17/spicy-burger-7323694_1280.jpg")
image2 = load_image("https://cdn.pixabay.com/photo/2020/01/28/12/52/burger-4799767_1280.jpg")

image1 = image1.resize((320, 320))
image2 = image2.resize((320, 320))

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(image1)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(image2)
plt.axis("off")

plt.show()


# Initialize processor and model
processor = AutoProcessor.from_pretrained("HuggingFaceTB/SmolVLM-Instruct")
model = AutoModelForImageTextToText.from_pretrained(
    "HuggingFaceTB/SmolVLM-Instruct",
    dtype=torch.bfloat16 if DEVICE == "cuda" else torch.float32,  # CPU에는 bfloat16 전용 연산기가 없어 float32보다 오히려 느려요 (GPU에서만 bfloat16)
    _attn_implementation="eager",
).to(DEVICE)

# Create input messages
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image"},
            {"type": "image"},
            {"type": "text",

             "text": "Can you explain the difference between the two images?"}
        ]
    },
]

# Prepare inputs
prompt = processor.apply_chat_template(messages, add_generation_prompt=True)
inputs = processor(text=prompt, images=[image1, image2], return_tensors="pt")
inputs = inputs.to(DEVICE)

# Generate outputs
generated_ids = model.generate(**inputs, max_new_tokens=500)
generated_texts = processor.batch_decode(
    generated_ids,
    skip_special_tokens=True,
)

print(generated_texts[0])
```
