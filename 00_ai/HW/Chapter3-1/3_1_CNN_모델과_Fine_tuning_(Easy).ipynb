### AlexNet의 등장 (사물인식)
- 2012년 제프리 힌튼 교수팀(토론토대)의 알렉스 크리제브스키가 AlexNet이라는 이름으로 ImageNet 대회에 나가 압도적 1위를 달성했습니다.
- 1980년대 개발된 CNN(Convolutional Neural Network)와 GPU를 활용해서 만든 모델입니다.
- AlexNet은 이런 의미가 있습니다.
  - Vision 연구자들에게 Neural Network (신경망) 활용 연구의 가능성을 입증해줌
  - Neural Network 연구가 활성화 된 계기 - 딥러닝 시대를 열었습니다.
  - Model 학습시 GPU 사용이 필수가 된 계기

### AlexNet 이후 사물인식의 큰 축
- ResNet (Residual Network, 2015년, Microsoft)
  - 발음 : 레즈넷
  - **CNN 기반에서 개선한 구조**
  - 당시 NN에서 Layer 층수가 깊어질수록, 오히려 성능이 떨어진다는 이슈가 있었습니다.
  - ResNet은 Layer수를 152층까지 연결하여 ImageNet에서 1위를 차지했습니다.
  - Layer를 늘려 더 좋은 성능을 낼수 있음을 입증하였고, 이후 CNN은 대부분 ResNet 기반으로 만들어집니다. (YOLO, EfficientNet 등)


- ViT (Vision Transformer, 2020년, Google)
  - 발음 : 브이 아이 티
  - **ViT는 CNN 구조가 아닙니다. Transformer 구조입니다.**
  - Transformer를 Vision에 적용하여, ImageNet에서 1위를 차지했습니다.
  - Transformer 구조가 Image에서도 쓸 수 있다는 것을 입증했습니다.
  - 이후 사물인식은 Tranformer가 대세가 되었습니다.

# 4. CNN 코드리뷰 (MNIST)

- 그럼 코드를 살펴보겠습니다.
- Convolution Block 에서 출력 채널의 의미는 제작할 Feature Map의 개수를 뜻합니다.
  - 하이퍼파라미터입니다.

````python
# 로컬 환경에서 sample_data 폴더에 MNIST CSV 파일이 없을 경우, 아래 코드로 다운로드 받습니다.
# (1_2 수업에서 사용한 것과 동일한 데이터입니다.)
import os
import urllib.request

os.makedirs('sample_data', exist_ok=True)

# (Google Machine Learning Crash Course 저장소 출처)
base_url = 'https://raw.githubusercontent.com/spiderPan/Google-Machine-Learning-Crash-Course/master/data/'
for file_name in ['mnist_train_small.csv', 'mnist_test.csv']:
    file_path = 'sample_data/' + file_name
    if not os.path.exists(file_path):
        print(file_name + ' 다운로드 중...')
        urllib.request.urlretrieve(base_url + file_name, file_path)
print('MNIST 데이터 준비 완료')
````

````python
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'사용 장치: {device}')

# ==================================================================================
# 1. 데이터 준비
# ==================================================================================
# csv 파일 읽기
print('데이터 준비중...', end='')
train_df = pd.read_csv("sample_data/mnist_train_small.csv")

# 0 ~ 1 정규화
X_train = torch.tensor(train_df.iloc[:, 1:].values, dtype=torch.float32) / 255.0
y_train = torch.tensor(train_df.iloc[:, 0].values, dtype=torch.long)

# CNN에 맞게 행렬로 변경
# -1 : 자동으로 이미지 수 채워짐
X_train = X_train.reshape(-1, 1, 28, 28) # (이미지 수, 채널, 높이, 가로),

# DataLoader 준비
# Test는 Shuffle이 필요없음
train_loader = DataLoader(list(zip(X_train, y_train)), batch_size=64, shuffle=True)
print('완료')


# ==================================================================================
# 2. 모델 준비
# ==================================================================================
print('모델 준비중...', end='')
model = nn.Sequential(
    # Conv Block 1
    nn.Conv2d(1, 32, kernel_size=5, padding=2), # (1,28,28) -> (32,28,28) 32개 피쳐맵 생성
    nn.ReLU(),
    nn.MaxPool2d(2, 2), # (32,14,14)

    # Conv Block 2
    nn.Conv2d(32, 64, kernel_size=5, padding=2), # (32,14,14) -> (64,14,14)
    nn.ReLU(),
    nn.MaxPool2d(2, 2), # (64,7,7)

    # Flatten
    nn.Flatten(), # (64,7,7) -> (3136)

    # FC (최종출력 = 10개 Layer)
    nn.Linear(3136, 128),
    nn.ReLU(),
    nn.Linear(128, 10)
).to(device) # [for GPU 1/3] GPU로 모델 이동

criterion = nn.CrossEntropyLoss() # Cross Entropy
optimizer = optim.Adam(model.parameters(), lr=0.001) # 아담
print('완료')

# ==================================================================================
# 3. 학습 시작
# ==================================================================================
print('학습 시작')
for epoch in range(3): # 3번 epoch만 수행
    print(f' - Epoch [{epoch+1}/3] ...', end='')
    running_loss = 0.0

    for images, labels in train_loader:
        # [for GPU 2/3] 입력 데이터를 GPU로 이동
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f' Loss : {running_loss/len(train_loader):.4f}')

print("학습 완료")
````

````python
import matplotlib.pyplot as plt

# ==================================================================================
# 1. 데이터 준비
# ==================================================================================
# csv 파일 읽기
test_df  = pd.read_csv("sample_data/mnist_test.csv")

# 0 ~ 1 정규화
X_test  = torch.tensor(test_df.iloc[:, 1:].values, dtype=torch.float32) / 255.0
y_test  = torch.tensor(test_df.iloc[:, 0].values, dtype=torch.long)

# CNN에 맞게 행렬로 변경
X_test  = X_test.reshape(-1, 1, 28, 28) # (이미지 수, 채널, 높이, 가로)

# DataLoader 준비
test_loader  = DataLoader(list(zip(X_test,  y_test)),  batch_size=3, shuffle=True)


# ==================================================================================
# 2. 모델에 입력 데이터 이미지 3장 선택
# ==================================================================================
images = None
for X_batch, y_batch in test_loader:
  images = X_batch
  break

# 입력할 이미지 출력
print('입력 이미지')
for i in range(3):
    plt.subplot(1, 3, i+1)
    record = images[i].reshape(28, 28);
    plt.imshow(record, cmap="gray")
    plt.axis("off")
plt.show()
print()

# ==================================================================================
# 3. 추론하기!
# ==================================================================================
with torch.no_grad():
    images = images.to(device) # GPU로 이미지 보내기
    outputs = model(images)
    preds = torch.argmax(outputs, dim=1) # 10개의 output 값중 가장 큰 값의 index 찾기
    print(f'모델 추론 결과 : {preds}')

````

# 5. TorchVision

- 이제부터 PyTorch 팀에서 만든 TorchVision 을 사용해서 실습을 할겁니다.
- TorchVision 이란?

# 6. AlexNet 모델
- AlexNet 모델은 앞에 수업했던 CNN을 디테일하게 튜닝했습니다.
  - 기존 CNN에 다양한 아이디어를 적용했지만, 큰 차별점은 다음과 같습니다.
  - 구조적으로 여러 Convolution Block을 사용했습니다.
  - GPU 을 이용한 학습을 했습니다.
  - 추가로 다양한 아이디어를 적용했습니다. (Dropout, 데이터 증강 등)

- AlexNet의 구조는 다음과 같습니다.
  - 5개의 Convolution 층 (모두 뒤에 ReLU)
    - 이 중 1, 2, 5번째 뒤에만 MaxPool이 붙습니다. (MaxPool은 총 3개)
  - Flatten
  - 3개의 FC

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_1_cnn_fine_tuning/22.png)

### AlexNet 추론 코드
- 다음은 학습이 완료된 AlexNet 모델을 다운로드 받고, 추론을 시켜보는 코드입니다.
- TorchVision Library를 처음 사용합니다. 코드를 모두 이해해주세요.

````python
import torch
from torchvision import models, transforms
from PIL import Image
from io import BytesIO
from torchvision.models import AlexNet_Weights
import requests
import matplotlib.pyplot as plt


# 모델 불러오기
# Alexnet은 옵션이 IMAGENET 1K version1 이거 하나밖에 없어요.
# ImageNet 128만개의 이미지로 학습된 1000개의 클래스를 분류하는 모델을 다운로드 받습니다.
model = models.alexnet(weights="IMAGENET1K_V1")

# 평가 모드로 전환
# 이제부터 추론일때와 학습할때 내부 수식이 달라집니다.
# 아래 코드를 수행하면 PyTorch 내부가 추론 수식으로 전환됩니다.
# (자세한 내용은 DropOut 이라는 기술을 참고해야하지만, 생략합니다.)
model.eval()

# 이미지 URL
#url = "https://cdn.pixabay.com/photo/2018/07/07/21/29/cat-3522912_1280.jpg" # 고양이
url = "https://cdn.pixabay.com/photo/2012/02/27/17/00/ape-17474_1280.jpg" #침팬지
#url = "https://cdn.pixabay.com/photo/2020/02/03/07/40/cross-speed-4814978_1280.jpg" #스포츠카

# 이미지 다운로드
img = Image.open(BytesIO(requests.get(url).content)).convert("RGB")

# 이미지 출력
plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.axis('off')
plt.show()

# 이미지 전처리
preprocess = transforms.Compose([
    transforms.Resize((256, 256)), # 사이즈 변경 256 x 256 픽셀
    transforms.ToTensor(), # 이미지 객체를 PyTorch의 Tensor로 자료형 변경 + 0 ~ 1 로 정규화
    transforms.Normalize(mean=[0.485, 0.456, 0.406], # RGB 채널별 평균과 표준편차
                         std=[0.229, 0.224, 0.225]), # 이 수치들은 모델이 학습할 때 사용한 정규화된 평균과 표준편차 값 그대로 사용
])

input_tensor = preprocess(img) # 차원 : (3, 256, 256) -> (채널, Height, Width)
input_tensor = input_tensor.unsqueeze(0) # 차원 : (1, 3, 256, 256) -> (배치, 채널, Height, Width)

# 추론
with torch.no_grad():
    outputs = model(input_tensor)
    preds = torch.argmax(outputs, dim=1).item()

# 결과 출력
labels = AlexNet_Weights.IMAGENET1K_V1.meta["categories"]
print(f'예측 클래스: {labels[preds]}')
````

- 모델이 학습할 때 사용했던 정규화 값입니다. (이미 알려진 값)

- 정규화값은 (평균, 표준편차)로 총 3개입니다. R, G, B 채널별 입니다.

      - [[1, 2]] 이 행렬은 shape가 (1, 2) 입니다.

# 7.  ResNet 모델
- ResNet은 Microsoft에서 CNN 에다가 다음과 같은 것을 추가하였습니다.
  - 잔차 연결 (Residual Connection) 이라는 수식을 추가
  - 더 깊은 Layer를 사용

- 2015년에 아주 높은 성능으로 ImageNet에서 우승했습니다.
  - 아래 차트에서 5% 빨간 휴먼은, 사람이 인식하는 수준입니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_1_cnn_fine_tuning/23.png)

  - 기존 CNN은 Layer가 많아질수록, Backpropagation 수식이 점점 0 으로 수렴해지는 문제가 있습니다.
    - GD는 기울기를 이용해서 얼마나 내려갈지 알게되는 원리입니다.
    - 그런데 기울기 계산 결과가 자꾸 0이 되어버리면 학습이 멈춰버립니다.
    - 이를 **기울기 소실(Vanishing Gradient)** 문제라고 합니다.
- 잔차 연결이라는 수식 덕분에 CNN 에서 더 깊은 Layer를 쓸 수 있게 되었습니다
  - 그냥 이전 레이어의 입력 값을 현재 레이어 출력값에 덧셈해줍니다.

- 현재 CNN을 사용한다고 하면, Default가 ResNet구조입니다.
- ResNet은 Level에 따라 아래와 같은 종류가 있습니다.
  - ResNet18 : 18층
  - ResNet50 : 50층
  - ResNet152 : 152층
  - 등등..

- ResNet이 가장 간단해서, ResNet-18로 수업하기에 좋습니다.
  - 아래 그림에서 "3 x 3 conv, 64" 에서 64는 출력 개수입니다. (Feature Map 개수)
  - 17개의 Convolution Layer + 1개의 FC로 구성되어있습니다.
  

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_1_cnn_fine_tuning/24.png)

### ResNet 추론 코드

- Torch Vision을 사용하면, AlexNet 코드와 차이가 없습니다.

````python
import torch
from torchvision import models, transforms
from PIL import Image
from io import BytesIO
from torchvision.models import ResNet18_Weights
import requests
import matplotlib.pyplot as plt

# 모델 불러오기
# 사전학습된 ResNet-18 모델을 불러옵니다.
model = models.resnet18(weights=ResNet18_Weights.IMAGENET1K_V1)
model.eval() # 평가모드 (추론용)

# 이미지 URL
url = "https://cdn.pixabay.com/photo/2018/07/07/21/29/cat-3522912_1280.jpg" # 고양이
#url = "https://cdn.pixabay.com/photo/2012/02/27/17/00/ape-17474_1280.jpg" #침팬지
#url = "https://cdn.pixabay.com/photo/2020/02/03/07/40/cross-speed-4814978_1280.jpg" #스포츠카

# 이미지 다운로드
img = Image.open(BytesIO(requests.get(url).content)).convert("RGB")

# 이미지 출력
plt.figure(figsize=(5, 5))
plt.imshow(img)
plt.axis('off')
plt.show()

# 이미지 전처리
preprocess = transforms.Compose([
    transforms.Resize((256, 256)), # 크기 변경 (256 x 256)
    transforms.ToTensor(), # PyTorch Tensor 타입으로 변경 + 0 ~ 1로 정규화
    transforms.Normalize( # 정규화
        mean=[0.485, 0.456, 0.406], # ImageNet 학습때 쓴 평균 값
        std=[0.229, 0.224, 0.225] # ImageNet 학습때 쓴 표준편차 값
    ),
])

# 이미지 전처리 후 차원 변경
# (3, 256, 256) -> (1, 3, 256, 256) : 배치차원 추가
input_tensor = preprocess(img).unsqueeze(0)

# 모델 추론
with torch.no_grad():
    outputs = model(input_tensor)
    preds = torch.argmax(outputs, dim=1).item()

# 예측 결과 출력
labels = ResNet18_Weights.IMAGENET1K_V1.meta["categories"]
print(f'예측 클래스: {labels[preds]}')
````

# 8. Fine-tuning 에 대해서
> 첨언: 파인튜닝은 **전이학습**(Transfer Learning, 한 작업에서 학습한 지식을 다른 작업으로 활용하는 것)을 구현하는 방법 중 하나 입니다.
- 파인튜닝은 이미 훈련된 모델에, 새로운 데이터를 추가로 학습시키는 방법입니다.
  - 파인튜닝을 왜 해야하는 가는, 너무나 당연해서 말할 필요가 없을 정도입니다. 불필요한 노력이 덜 들어가거든요.
  - 비유
    - (모델을 처음서부터 만들기) vs (세계 최고의 모델 + 적은 데이터로 추가 학습)
    - (웹프레임워크 없이 개발하기) vs (웹프레임워크 + 코드 추가로 개발하기)

- 파인튜닝의 범위로 다음과 같이 구분할 수 있습니다.
  1. Full Fine-tuning
      - 추가학습시, 모델 전체 파라미터가 변경되도록 합니다.
      - **성능이 잘 나올만큼 신규 데이터를 많이 구하기 어려워서, 이 방법은 사용하지 않습니다.**
  2. Partial Fine-tuning
      - 추가학습시, 일부 Layer의 파라미터가 변경되도록 합니다.
      - 데이터는 적게 필요하지만, 그 만큼 성능 향상이 제한적이고 많은 실험이 필요합니다.
      - 데이터를 많이 넣어도 그 만큼 성능이 잘 안나와서, 효율적인 파인튜닝 대비 가성비가 떨어집니다.
      - **가성비는 떨어지지만, 파인튜닝 개념을 익히기에는 좋습니다. ^^ 즉, 학습용!**
  3. 효율적인 Fine-tuning
      - 추가학습시, 모델 대부분을 변경하지 않고 추가 Layer에 학습 후 연결만 합니다.
      - Parameter-Efficient Fine-tuning(효율적 파인튜닝, PEFT) 라고부릅니다.
      - **현재 Fine-tuning 이라고 말하면, 보통 이 방식을 의미합니다.**
      - PEFT 는 별도의 수업에서 다룹니다.
      


- 이번 시간에는 Partial Fine-tuning 중 변경 범위를 가장 좁힌 방법으로 Fine-tuning을 할겁니다.
  - 미리 말씀드릴게요. 성능은 기대하지마세요^^

# 9. Linear Probing
- Partial Fine-tuning 중에 가장 변경 범위를 좁혀 사용하는 방법입니다.
- 마지막 Linear Layer의 가중치만 초기화 후, 새롭게 학습시킵니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_1_cnn_fine_tuning/25.png)

- 마지막에 단 1개의 FC가 있습니다. 입력이 512개, 출력이 1000개 입니다.
  - 512개는 저 그림에 써있어요~ Resnet-18 구조가 그래요.
  - 출력은 1,000개 입니다. ImageNet 데이터의 클래스가 1,000개 예요.
  - 이 FC의 역할은 그동안 Feature Map의 최종 결과물을, 1000개의 클래스로 분류해주는 역할을 담당합니다.

### 파인튜닝으로 만들고자 하는 것
- 현재 1,000개의 클래스를 분류하는데 성공시킨 멋진 특징 추출기가 있습니다.
- 특징추출기를 Backbone(백본, 척추) 라고 합니다.
- 이 척추를 그대로 사용하고, FC만 새로 바꿔 다른 분류기를 만들고자 합니다. 강아지 2진 분류기입니다.

![image.png](https://raw.githubusercontent.com/withSG/ai_assets/master/14_easy/3_1_cnn_fine_tuning/26.png)

### 파인튜닝 방법
- 기존 FC는 백본으로 부터 입력 512개를 받고, 1,000개의 클래스로 분류해줍니다.
- 신규 FC는 백본으로 부터 입력 512개를 받고, 2개의 클래스로 분류해줄겁니다.
  - 사진이 강아지인지 아닌지를 확률로 계산합니다.

- 핵심 코드는 다음과 같습니다.
  - 가중치값 변경 못하도록, 자동 미분 계산 기능 OFF 합니다.
  - 그러면 Backpropagation시 기울기가 계산되지 않아, 가중치가 업데이트 되지 못합니다.

```python
# [핵심코드 1] 모든 파라미터 autograd off
for param in model.parameters():
    param.requires_grad = False

# [핵심코드 2] 마지막 FC 변경 (새로 생성)
model.fc = nn.Linear(512, 2) # 입력 512개, 출력 2개, 이 부분만 autograd on
```

````python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, datasets, transforms
from torch.utils.data import DataLoader
import os


# =============================================================================
# 1. 기본 설정
# =============================================================================

# 로컬에서만 실행 (실행 후 Restart 필수)
#!pip uninstall torch torchvision torchaudio -y
#!pip uninstall torch torchvision torchaudio -y
#!pip install -q "torch>=2.1.0" "torchvision>=0.16.0"
#!nvidia-smi
#print("CUDA 사용 가능:", torch.cuda.is_available())
#print("GPU 이름:", torch.cuda.get_device_name(0))

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("사용 장치 : ", device)

# 시드 값 고정 (항상 같은 결과로 학습되기 위해)
# 첫 학습 시, 할당할 모델의 파라메터 값을 랜덤으로 결정하는데
# 동일한 학습을 위해 시드 값을 고정
SEED = 0
torch.manual_seed(SEED)


# =============================================================================
# 2. 이미지 전처리
# =============================================================================
print('이미지 준비...', end='')
transform_train = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

# train 폴더에 이미지들 정보를 읽음 (폴더명을 클래스명으로 인식!)
train_dataset = datasets.ImageFolder("dog_image/train", transform=transform_train)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
print('완료')


# =============================================================================
# 3. 사전학습된 ResNet18 모델 준비 + 파인튜닝 준비
# =============================================================================
print('모델 준비...', end='')
model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)

# [핵심코드 1] 모든 파라미터 autograd off
for param in model.parameters():
    param.requires_grad = False

# [핵심코드 2] 마지막 FC 변경 (새로 생성)
model.fc = nn.Linear(512, 2) # 입력 512개, 출력 2개, 이 부분만 autograd on
model = model.to(device)
print('완료')


# =============================================================================
# 4. 학습
# =============================================================================
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.fc.parameters(), lr=0.001)

epochs = 10
print('학습 시작')
for epoch in range(1, epochs+1):  # 🔹 Epoch 수는 자유롭게 조절 가능
    print(f' - [Epoch {epoch}/{epochs}] ... ',end='')
    model.train()
    total_loss = 0.0

    for imgs, labels in train_loader:
        imgs, labels = imgs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)
    print(f'Loss: {avg_loss:.4f}')
print('학습 완료')
````

- 학습이 완료되었으니 추론을 해봅시다.

````python
import torch
from torchvision import transforms
from PIL import Image
from io import BytesIO
import requests
import matplotlib.pyplot as plt

# 테스트 해볼 이미지 URL들
urls = [
    "https://cdn.pixabay.com/photo/2012/11/28/10/28/cowboy-67630_1280.jpg",  # 사람
    "https://cdn.pixabay.com/photo/2015/11/17/13/13/puppy-1047521_1280.jpg",  # 강아지
    "https://cdn.pixabay.com/photo/2018/07/07/21/29/cat-3522912_1280.jpg",  # 고양이
    "https://cdn.pixabay.com/photo/2012/02/27/17/00/ape-17474_1280.jpg",  # 침팬지
    "https://cdn.pixabay.com/photo/2021/07/05/14/07/dog-6389277_1280.jpg",  # 강아지
    "https://cdn.pixabay.com/photo/2020/02/03/07/40/cross-speed-4814978_1280.jpg",  # 스포츠카
    "https://cdn.pixabay.com/photo/2020/01/21/16/26/yorkshire-terrier-4783327_1280.jpg",
    "https://cdn.pixabay.com/photo/2013/06/20/04/41/fool-140229_1280.jpg",
    "https://cdn.pixabay.com/photo/2021/10/10/21/52/makeup-6698881_1280.jpg",
    "https://cdn.pixabay.com/photo/2019/11/04/01/11/cellular-4599956_1280.jpg",
]

results = []
model.eval()

with torch.no_grad():
    for url in urls: # 이미지 하나씩 추론해봄

        # 다운로드 1 장
        img = Image.open(BytesIO(requests.get(url).content)).convert("RGB")

        # 이미지 전처리
        img_tensor = transform_train(img).unsqueeze(0).to(device)

        # 추론
        outputs = model(img_tensor)

        # 확률계산
        probs = torch.softmax(outputs, dim=1)[0].cpu().numpy()

        # 결과를 result에 저장
        pred_idx = probs.argmax()
        pred_class = train_dataset.classes[pred_idx]
        pred_prob = probs[pred_idx]

        results.append({
            "img": img,
            "pred_prob": pred_prob,
            "pred_class": pred_class,
            "probs": probs
        })

# 'dogs' 클래스로 예측된 결과만 필터링
results = [r for r in results if r['pred_class'] == 'dogs']
print(f"총 이미지 {len(urls)}개 중에서 강아지로 예측된 이미지는 {len(results)}개 입니다.")

# 확률 높은 순서대로 정렬
results_sorted = sorted(results, key=lambda x: x["pred_prob"], reverse=True)

# 시각화
plt.figure(figsize=(15, 4))
for i, r in enumerate(results_sorted):
    plt.subplot(1, len(results_sorted), i + 1)
    plt.imshow(r["img"].resize((128, 128)))
    plt.axis("off")
    plt.title(f"Top-{i+1}\n{r['pred_prob']:.2f}")

print()
````

- 강아지일 확률이 가장 높은 사진 순서대로 출력하였습니다.
  - ResNet-18의 백본을 그대로 사용하면서, 이진 분류기로 파인튜닝을 해보았습니다.
  - 성능은 좋지 못하지만, 학습용으로 공부해주시면 됩니다.
